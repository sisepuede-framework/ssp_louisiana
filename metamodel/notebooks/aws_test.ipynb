```python
SCRIPT_DIR_PATH = os.getcwd()
ROOT_DIR_PATH = os.path.dirname(SCRIPT_DIR_PATH)
DATA_DIR_PATH = os.path.join(ROOT_DIR_PATH, "data")
SSP_DIR_PATH = os.path.join(DATA_DIR_PATH, "ssp")
CONFIG_DIR_PATH = os.path.join(ROOT_DIR_PATH, "config")
```

### NOTE: Before running this notebook make sure to login into your profile using aws sso from the terminal

```python
# Initialize class that help us read yamls
edau = EDAUtils()
```

```python
# Read AWS credentials and bucket name from the config file
config = edau.read_yaml(os.path.join(CONFIG_DIR_PATH, "aws_credentials_config.yaml"))
profile_name = config["profile_name"]
bucket_name = config["bucket_name"]
```

```python
# Set your profile
session = boto3.Session(profile_name=profile_name)

# Create an S3 client or resource
s3 = session.resource('s3')

# Define folder prefix (where our files are stored in S3)
prefix = 'sisepuede_run_2025-01-14T17;04;06.975301_output_database/'  # this is like the "folder" in S3
```

```python
# Local destination to save the downloaded files
destination = os.path.join(SSP_DIR_PATH, "aws_downloaded_data_test")
os.makedirs(destination, exist_ok=True)

# Download files from S3 bucket
bucket = s3.Bucket(bucket_name)
for obj in bucket.objects.filter(Prefix=prefix):
    if obj.key.endswith('/'):  # skip directories
        continue
    target_path = os.path.join(destination, os.path.basename(obj.key))
    bucket.download_file(obj.key, target_path)
    print(f"Downloaded: {obj.key}")
```
