```python
import os
import pandas as pd
import numpy as np
```

```python
# Set up paths
SCRIPT_DIR_PATH = os.getcwd()
CW_DIR_PATH = os.path.join(SCRIPT_DIR_PATH, "cw")
DATA_DIR_PATH = os.path.join(SCRIPT_DIR_PATH, "data")
OUTPUT_DIR_PATH = os.path.join(SCRIPT_DIR_PATH, "output")
```

```python
rescaled_df = pd.read_csv(os.path.join(OUTPUT_DIR_PATH, "sisepuede_results_sisepuede_run_2025-01-14T17;04;06.975301_IDE_WIDE_INPUTS_OUTPUTS.csv"))
rescaled_df.head()
```

```python
rescaled_df_filtered = rescaled_df[rescaled_df["time_period"] == 7]
rescaled_df_filtered
```

```python
rescaled_df_filtered = rescaled_df_filtered[rescaled_df_filtered["primary_id"].isin([354601, 354405, 355053])]
rescaled_df_filtered
```

```python
# filter to only have fields that start with "emission_co2e_"
rescaled_df_filtered = rescaled_df_filtered.filter(regex='^emission_co2e_')
rescaled_df_filtered
```

```python
# Columns to check (exclude 'primary_id')
cols_to_check = rescaled_df_filtered.columns.difference(['primary_id'])

# Find columns that have more than one unique value
diff_cols = [col for col in cols_to_check if rescaled_df_filtered[col].nunique() > 1]

if diff_cols:
    print("These columns have different values across rows:", diff_cols)
    # Get rows where at least one diff_col is different from the first row
    mask = (rescaled_df_filtered[diff_cols] != rescaled_df_filtered[diff_cols].iloc[0])
    differing_rows = rescaled_df_filtered[mask.any(axis=1)]
    print("Rows with differences:")
    display(differing_rows)
else:
    print("✅ All columns (except primary_id) have the same values across all rows.")
```
